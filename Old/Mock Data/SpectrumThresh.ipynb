```python
spec = spec[spec.sum(axis=1)>0.25*10*8]
```

```python
# let's make some fake data from the spectrum traces
fakeSpectrum = np.zeros_like(spec)

half_length = spec.shape[1] // 3  # Assuming spec is a 2D array

for i in range(spec.shape[0] - 1):
    left_padded = np.pad(spec[i], (half_length, 0), mode='constant')[:spec.shape[1]]
    right_padded = np.pad(spec[i+1], (0, half_length), mode='constant')[-spec.shape[1]:]
    fakeSpectrum[i] = left_padded + right_padded
```

```python
plt.plot(fakeSpectrum.mean(axis=0))
```

```python
integratedSpecCorrected= fakeSpectrum.sum(axis=1)
probe_int = fakeSpectrum[:,0:1024].sum(axis=1)
I2Ratio = probe_int / integratedSpecCorrected
I1Ratio = 1 - I2Ratio

plt.figure(figsize=[10,4])
plt.subplot(1,2,1)
plt.hist(I1Ratio,bins=100,range=[0,1]);
plt.title('I1 Ratio')
plt.xlabel('Ratio')
plt.ylabel('Counts')
plt.subplot(1,2,2)
plt.hist(I2Ratio,bins=100,range=[0,1]);
plt.title('I2 Ratio')
plt.xlabel('Ratio')
plt.ylabel('Counts')
plt.show()
```
