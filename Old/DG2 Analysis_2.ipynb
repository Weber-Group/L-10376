# Radial Analyzer.ipynb

Loads in individual azimuthally averaged Jungfrau frames and all other smalldata parameters such as X-ray pulse energy, timetool edge position, etc. Can be used to look at static scattering patterns, filter out bad shots, build time-resolved difference signals, and much more.

```python
# Finding the basline of the dg2 traces
dg2Baseline = np.mean(np.sum(dg2Trace, axis=1)[:, 0:50], axis=1)
dg2Sum = np.sum(-np.sum(dg2Trace, axis=1) + dg2Baseline[:, np.newaxis], axis=1)[xrayOn]
# Summing the azav and 
# Now plotting the correlation between the dg2 sum and the azav summed
plt.hist2d(dg2Sum,numPhotons[xrayOn],bins=100)
plt.xlabel('DG2 Reading')
plt.ylabel('J4M Sum')
plt.title('L-10350 Run 249')
plt.show()
```

```python
# Now to generate our data which we will plot against this, also filtering out the no xray shots
normAzav = azav[xrayOn]/numPhotons[xrayOn,np.newaxis]

# Define bin edges (you can adjust the number of bins)
num_bins = 7  # Adjust as needed
bin_edges = np.linspace(dg2Sum.min(), dg2Sum.max(), num_bins + 1)

# Digitize dg2Sum to get bin indices
bin_indices = np.digitize(dg2Sum, bin_edges) - 1  # Adjust for zero-based indexing

# Initialize an array to store averaged normAzav
binned_means = np.zeros((num_bins, normAzav.shape[1]))

# Compute mean normAzav for each bin
for i in range(num_bins):
    mask = bin_indices == i  # Get indices belonging to this bin
    if np.any(mask):  # Check if bin is not empty
        binned_means[i] = np.mean(normAzav[mask], axis=0)
        plt.plot(q,binned_means[i]*q)
plt.show()
```

```python
pdiff = np.zeros((num_bins, normAzav.shape[1]))
ref = 1
for i in range(num_bins):
    pdiff[i] = 100*(binned_means[i]-binned_means[ref])/binned_means[ref]
    plt.plot(q,pdiff[i])
plt.show()
```
