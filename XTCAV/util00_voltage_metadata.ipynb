```python
detNames = psana.DetNames()
print('*** Detector Names (Full-Name, DAQ-Alias, User-Alias) ***')
for detname in detNames: 
    print(detname)
```

```python
epicsNames = psana.DetNames('epics')
print('*** Some Epics Names (Full-Name, DAQ-Alias, User-Alias) ***')
for ename in epicsNames: 
    print(ename)
```

```python
# Get detectors with 'XTCAV' in ther names

xtcav_detNames = []
str_comp = ('xtcav','XTCAV')
for ename in epicsNames:
    for name in ename:
        for s in str_comp:
            if s in name:
                if ename not in xtcav_detNames:
                    xtcav_detNames.append(ename)
```

```python
# Print xtcav det names

for name in xtcav_detNames: 
    print(name)
```

```python
# get XTCAV short aliases

xtcav_detNames_short = []

for name in xtcav_detNames:
    xtcav_detNames_short.append(name[1])
```

```python
for i,n in enumerate(xtcav_detNames_short):
    print(f"   {i:<2}     {n}")
```

## XTCAV Voltage

The XTCAV user's guide indicates the voltage is stored as:

- `     Item        |            PV          |   Unit   |     Updates    |     Recorded as...  `
- `XTCAV Voltage   |    TCAV:DMP1:360:ABR   |    MV    |     120 Hz     |     BLD (EBeam v5)  `

```python
# Search for item

detNames_voltageSearch = []
str_comp = ('TCAV','BLD',)
for ename in epicsNames:
    for name in ename:
        for s in str_comp:
            if s in name:
                if ename not in detNames_voltageSearch:
                    detNames_voltageSearch.append(ename)

for i,name in enumerate(detNames_voltageSearch):
    print(f"  {i:<2}    {name}")
```

```python
detName = 'EBeam'

det = psana.Detector(detName,env)

print(det)
```

```python
dataSource=psana.DataSource("exp=cxil1037623:run=%d:idx"%runnum)
env = dataSource.env()
det = psana.Detector('EBeam',env)
run = next(dataSource.runs())
times = run.times()
t_idx = 0
evt = None
while evt is None and t_idx<1000:
    t = times[t_idx]
    evt = run.event(t)
    t_idx += 1

print(t_idx, evt)
```

```python
ebeam = det.get(evt)
```
