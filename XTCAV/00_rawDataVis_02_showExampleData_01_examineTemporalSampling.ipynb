# XTCAV: Show Example Data & Examine Temporal Pixel Density

```python
import numpy as np
import matplotlib.pyplot as plt
from xtcav_processor import XTCAVProcessor
import psana
get_datasource = lambda runnum: psana.DataSource(f"exp=cxil1037623:run={runnum}:idx")
```

## Dark reference example

```python
# Get and show some data

runnum = 75

print(f"Showing dark reference (data run #{runnum}) sample XTCAV camera image capture...")
print("--------------------------------------------------------------------------")
print()

# Create processor object
# get the first good event
xtpr = XTCAVProcessor(get_datasource(runnum))
i,j,evt = next(xtpr.shot_iterator)

print(f"In data run {runnum} 'good' event #{i+1} occurs at time index {j}.")
print("Setting the event...")
print()
print("-------------------------------------------")
print(f"Data run {runnum} event {j} XTCAV camera capture:")

# Set the event
# Show the current frame
xtpr.set_current_event(evt)
xtpr.show_current_frame(vrange=(0,1e2),bins=np.linspace(0,1000,100),zoom=False)
```

## Single pulse examples

```python
fov_singlepulse = 120,240
```

```python
# Get and show some data

runnum = 11
n_shots_show = 3

print(f"Showing a few single pulse (data run #{runnum}) sample XTCAV camera image captures...")
print("-----------------------------------------------------------------------------")
print()

# Create the processor object
xtpr = XTCAVProcessor(get_datasource(runnum))

print()
print(f"Looping through the first {n_shots_show} good events...")
for idx,(i,j,evt) in enumerate(xtpr.shot_iterator):
    print(f"Showing good shot #{i+1} at time index {j}:")

    # Get & show event
    xtpr.set_current_event(evt)
    xtpr.show_current_frame(vrange=(0,350),bins=np.linspace(0,1000,100),fov=fov_singlepulse)

    # continue?
    if idx>=n_shots_show-1:
        break
```

# Examine temporal pixel density

## ...for single pulse (run 11)

```python
# Get and show some data

runnum = 11

ybar_pos = 160,208

print(f"Showing a data run #{runnum} XTCAV camera image capture with temporal extent delimiters shown")
print("--------------------------------------------------------------------------------------------")
print()

# Create the processor & get an event
xtpr = XTCAVProcessor(get_datasource(runnum))
i,j,evt = next(xtpr.shot_iterator)

# Get & show event
print(f"Showing good shot #{i+1} at time index {j}:")
xtpr.set_current_event(evt)
fig,axs = xtpr.show_current_frame(vrange=(0,350),bins=np.linspace(0,1000,100),fov=fov_singlepulse,returnfig=True)

# vertical bars
ax1,ax2,ax3 = axs
ax1.vlines(ybar_pos,0,1023,color='w',lw=1)
ax2.vlines(ybar_pos,0,1024,color='w',lw=1)
plt.show()
```
