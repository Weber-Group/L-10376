# XTCAV: Raw Data Visualization
# Search for Reference Data
# Epochs D-J

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
from utils import (
    load_xtcav_run_first_pulse,
    get_com,
    get_xtcav_mask,
    get_com_clean,
    get_plot_lims
)
```

 # Epoch D
 # - Runs 121-150
 # - Refs 116 (dark), 117 (no lasing)

## Run 116 (Dark)

```python
# set the run number
runnum=116

# set histogram range
vmin = 0
vmax = 1e2
```

```python
# load data
image,success=load_xtcav_run_first_pulse(runnum)

# show camera capture
fig,ax = plt.subplots()
ax.matshow(image,vmin=vmin,vmax=vmax)
ax.invert_yaxis()
plt.show()

# show histogram
fig,ax = plt.subplots()
ax.hist(image.ravel(),bins=np.linspace(0,1000,100))
ax.semilogy()
plt.show()
```

## Runs 117-125

```python
# Set parameters

# run numbers
runnums = np.arange(117,125)

# zoom-in FOV
Lx = 150
Ly = 240

# contrast
vmin,vmax = 0,5e2
```

```python
# Populate data

# set up storage
s = image.shape
n = (len(runnums))
ims = np.empty((n,s[0],s[1]))
coms = np.empty((n,2))

# get data
for idx,runnum in enumerate(runnums):
    # load
    image,success=load_xtcav_run_first_pulse(runnum)
    # get CoM
    com = get_com_clean(image)
    # store
    ims[idx] = image
    coms[idx] = com

# show
fig,axs = plt.subplots(2,4,figsize=(11,8))
for idx,runnum in enumerate(runnums):
    # get axis
    axi = idx//4
    axj = idx%4
    ax = axs[axi,axj]
    # get data
    im = ims[idx]
    com = coms[idx]
    # show
    ax.matshow(im,vmin=vmin,vmax=vmax)
    ax.set_xlim(get_plot_lims(com[1],Lx))
    ax.set_ylim(get_plot_lims(com[0],Ly))
    # run numbers
    ax.text(0.03,0.97,f"{runnums[idx]}",size=16,color='w',ha='left',va='top',transform=ax.transAxes)

plt.show()
```
