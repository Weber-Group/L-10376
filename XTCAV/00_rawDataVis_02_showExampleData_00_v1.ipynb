## Set up

```python
# Contrast ranges
vlims = vmin,vmax = 0,5e2
vlims_dark = vmin_dark,vmax_dark = 0,1e2

# zoom-in FOV
fovs = Lx,Ly = 150,240
```

```python
# utility
def add_ref_color_spines(n_ref, n_first, fig, axs, color_ref='tomato', color_first='dodgerblue', lw=3):
    ## (1) reference
    # get axis
    axi = n_ref//4
    axj = n_ref%4
    ax = axs[axi,axj]
    for p in ('top','bottom','right','left'):
        ax.spines[p].set_edgecolor(color_ref)
        ax.spines[p].set_linewidth(lw)
    ## (2) first
    # get axis
    axi = n_first//4
    axj = n_first%4
    ax = axs[axi,axj]
    for p in ('top','bottom','right','left'):
        ax.spines[p].set_edgecolor(color_first)
        ax.spines[p].set_linewidth(lw)
    pass
```
