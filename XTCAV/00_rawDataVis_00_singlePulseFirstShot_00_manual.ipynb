```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import psana
from xtcav.ShotToShotCharacterization import ShotToShotCharacterization
import warnings
```

## Run 10

```python
# set the run number
runnum=10

# data source
dataSource=psana.DataSource("exp=cxil1037623:run=%d:idx"%runnum)
epicsStore=dataSource.env().epicsStore()

# XTCAV data shot-by-shot retrieval object
XTCAVRetrieval=ShotToShotCharacterization();
XTCAVRetrieval.SetEnv(dataSource.env())

# get the run and shot timing data
run = next(dataSource.runs())
times = run.times()

# find the first event with XTCAV data
t_idx = 0
success = False
while not success and t_idx<1000:
    t = times[t_idx]
    evt = run.event(t)
    success = XTCAVRetrieval.SetCurrentEvent(evt)
    #print(f"t={t_idx}: success = {success}")
    t_idx += 1

# get the XTCAV camera raw image data
image,ok0=XTCAVRetrieval.RawXTCAVImage()

# show
fig,ax = plt.subplots()
ax.matshow(image)
ax.invert_yaxis()
plt.show()
```

```python
fig,ax = plt.subplots()
ax.hist(image.ravel(),bins=np.linspace(0,1000,100))
# ax.set_ylim(0,4e5)
ax.semilogy()
plt.show()
```

```python
fig,ax = plt.subplots()
ax.matshow(image,vmin=0,vmax=5e2)
ax.set_xlim(200,440)
ax.set_ylim(300,640)
plt.show()
```
